```python
import pandas as pd

# Load each sheet separately
orders = pd.read_excel("Raw Data.xlsx", sheet_name="orders")
customers = pd.read_excel("Raw Data.xlsx", sheet_name="customers")
products = pd.read_excel("Raw Data.xlsx", sheet_name="products")

print(orders.columns)
print(customers.columns)
print(products.columns)
```

```python
# Merge orders with customers using Customer ID
df = orders.merge(customers[['Customer ID', 'Customer Name', 'Email', 'Country', 'Loyalty Card']], 
                  on='Customer ID', 
                  how='left')

# Merge result with products using Product ID
df = df.merge(products[['Product ID', 'Coffee Type', 'Roast Type', 'Size', 'Unit Price', 'Profit']], 
              on='Product ID', 
              how='left')

df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
# Drop the empty _x columns and rename _y columns
df = df.drop(columns=['Customer Name_x', 'Email_x', 'Country_x', 
                       'Coffee Type_x', 'Roast Type_x', 'Size_x', 
                       'Unit Price_x', 'Sales'])

df = df.rename(columns={
    'Customer Name_y': 'Customer Name',
    'Email_y': 'Email',
    'Country_y': 'Country',
    'Coffee Type_y': 'Coffee Type',
    'Roast Type_y': 'Roast Type',
    'Size_y': 'Size',
    'Unit Price_y': 'Unit Price'
})

# Calculate Sales column
df['Sales'] = df['Quantity'] * df['Unit Price']

print(df.columns)
print(df.isnull().sum())
```

```python
# 1. Which coffee type sells most?
df.groupby('Coffee Type')['Sales'].sum().sort_values(ascending=False)
```

```python
# 2. Which country generates most revenue?
df.groupby('Country')['Sales'].sum().sort_values(ascending=False)
```

```python
# 3. Monthly sales trend
df['Month'] = df['Order Date'].dt.to_period('M')
df.groupby('Month')['Sales'].sum()
```
